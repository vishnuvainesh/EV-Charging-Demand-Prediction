<h3 style="color:white">Convert Datetime Columns</h3>

```python
df["StartDatetime"] = pd.to_datetime(df["StartDatetime"])
df["EndDatetime"] = pd.to_datetime(df["EndDatetime"])

df[["StartDatetime", "EndDatetime"]].dtypes
```

<h3 style="color:white">Check Negative Duration</h3>

```python
df[df["Duration"] < 0][["StartDatetime", "EndDatetime", "Duration"]]
```

<h3 style="color:white">Recalculate Duration</h3>

```python
df["Duration"] = (
    df["EndDatetime"] - df["StartDatetime"]
).dt.total_seconds() / 60

df["Duration"] = df["Duration"].astype(int)

df["Duration"].describe()
```

<h3 style="color:white">Remove Invalid Duration Records</h3>

```python
df = df[df["Duration"] >= 0].copy()

df.shape
```
